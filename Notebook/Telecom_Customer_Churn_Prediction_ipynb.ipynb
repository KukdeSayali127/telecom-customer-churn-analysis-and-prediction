```python
# STEP 7: Encoding (Machine Learning Preparation)

df_ml = df.copy()

# Convert Churn to numbers
df_ml["Churn"] = df_ml["Churn"].map({"Yes": 1, "No": 0})

# Convert all text columns into numbers
df_ml = pd.get_dummies(df_ml, drop_first=True)

df_ml.head()
```

```python

# STEP 7 : Encode Categorical Columns


df_ml = df.copy()

# Convert TotalCharges into numeric
df_ml["TotalCharges"] = pd.to_numeric(df_ml["TotalCharges"], errors="coerce")
df_ml["TotalCharges"] = df_ml["TotalCharges"].fillna(df_ml["TotalCharges"].median())

# Convert Churn into numbers
df_ml["Churn"] = df_ml["Churn"].map({"Yes":1, "No":0})

# Convert all categorical columns into numbers
df_ml = pd.get_dummies(df_ml, drop_first=True)

print(df_ml.head())
```

```python

# STEP 8 : Features and Target


X = df_ml.drop("Churn", axis=1)   # Input Features

y = df_ml["Churn"]                # Target Column

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python

# STEP 9 : Train-Test Split


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Data :", X_train.shape)
print("Testing Data  :", X_test.shape)
```

```python

# STEP 10 : Random Forest Model


from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

prediction_rf = rf_model.predict(X_test)

print("Random Forest Model Trained Successfully!")
```

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(
    max_iter=1000,
    solver="liblinear",
    random_state=42
)

lr_model.fit(X_train, y_train)

prediction_lr = lr_model.predict(X_test)

print("Logistic Regression Model Trained Successfully!")
```

```python

# STEP 12 : Accuracy Score


from sklearn.metrics import accuracy_score

rf_accuracy = accuracy_score(y_test, prediction_rf)
lr_accuracy = accuracy_score(y_test, prediction_lr)

print("Random Forest Accuracy :", rf_accuracy)
print("Logistic Regression Accuracy :", lr_accuracy)
```

```python

# STEP 14 : Confusion Matrix


from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, prediction_lr)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap="Blues")

plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python

# STEP 15 : Classification Report


from sklearn.metrics import classification_report

print(classification_report(y_test, prediction_lr))
```

```python

# STEP 16 : Save Final Dataset for Power BI


results = X_test.copy()

results["Actual_Churn"] = y_test.values
results["Predicted_Churn"] = prediction_lr   # Logistic Regression prediction

results.to_csv("Customer_Churn_Predictions.csv", index=False)

print("File Saved Successfully!")
```
